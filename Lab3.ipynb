```python
from scipy.stats import norm
from math import sqrt

n = 1710
total = 1586500
mu0 = 500
sigma = sqrt(250)

#Зробимо Z-тест, цікаво тема підібрана

x_mean = total / n

z = n**0.5 * (x_mean - mu0) / sigma
p_value = 2 * norm.sf(abs(z))

print("Середнє:", x_mean)
print("Z-stat:", z)
print("P-value:", p_value)

if p_value < 0.05:
    print("Відхиляємо H0")
else:
    print("Приймаємо H0")
```
